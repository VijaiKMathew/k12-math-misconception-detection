  <h2 style="margin: 0; font-size: 20px; font-weight: 800; color: #ffffff;">2. 🎯 Mathematical Metric Derivation: Demystifying MAP@3</h2>
  <p style="margin: 4px 0 0 0; font-size: 13px; color: #94a3b8;">Exact reciprocal-rank decomposition and why Rank #1 vs Rank #2 is a +0.50 step function</p>
</div>

Because each student trace $i \in \{1, \dots, N\}$ has **exactly one** ground-truth root label $y_i \in \mathcal{C} \cup \{	ext{NONE}\}$, Mean Average Precision @ 3 ($	ext{MAP@3}$) simplifies to the mean reciprocal rank truncated at cutoff $K=3$:

$$\text{MAP@3} = \frac{1}{N} \sum_{i=1}^{N} \left( \mathbf{1}[\hat{c}_{i,1} = y_i] \cdot 1.0 \;+\; \mathbf{1}[\hat{c}_{i,2} = y_i] \cdot 0.5 \;+\; \mathbf{1}[\hat{c}_{i,3} = y_i] \cdot \frac{1}{3} \right)$$

<div style="background: #f8fafc; border: 1px solid #cbd5e1; border-left: 4px solid #0284c7; border-radius: 8px; padding: 14px 18px; margin: 12px 0; color: #0f172a;">
  <b>💡 Two Strategic Laws of MAP@3 Optimization:</b>
  <ol style="margin: 6px 0 0 18px; font-size: 13.5px; line-height: 1.6;">
    <li><b>Candidate Horizon Law:</b> Any trace whose true label $y_i$ is excluded from the candidate pool $\mathcal{H}(i)$ receives a hard <code>0.0</code> regardless of how strong the downstream reranker is.</li>
    <li><b>Sibling Swap Leverage:</b> Promoting the true misconception from Rank #2 to Rank #1 doubles its score contribution from <code>0.50</code> to <code>1.00</code> (<code>+0.50</code> net gain per trace).</li>
  </ol>
</div>

```python
from __future__ import annotations
import math
import numpy as np
import pandas as pd

def compute_map3(actual: list[str], predicted: list[list[str] | str]) -> float:
    """Exact vectorized-ready MAP@3 metric with duplicate-prediction guard."""
    if not actual:
        return 0.0
    total = 0.0
    for target, preds in zip(actual, predicted):
        tokens = preds.split() if isinstance(preds, str) else list(preds)
        seen: list[str] = []
        for tok in tokens:
            if tok not in seen:
                seen.append(tok)
            if len(seen) == 3:
                break
        for rank, p in enumerate(seen):
            if p == target:
                total += 1.0 / (rank + 1)
                break
    return total / len(actual)

# Verify exact scoring behavior across Ranks 1, 2, 3, and Miss
demo_cases = [
    ("Rank #1 Exact Match", ["KPF-01"], ["KPF-01 KPF-02 CAL-01"]),
    ("Rank #2 Sibling Hit", ["KPF-01"], ["KPF-02 KPF-01 CAL-01"]),
    ("Rank #3 Backup Hit ", ["KPF-01"], ["KPF-02 CAL-01 KPF-01"]),
    ("Outside Top-3 Miss ", ["KPF-01"], ["NONE ANS-01 CAL-01"]),
]
print("=" * 62)
print(f"{'Scenario':<25} | {'Prediction':<22} | {'MAP@3':>8}")
print("-" * 62)
for label, act, pred in demo_cases:
    print(f"{label:<25} | {pred[0]:<22} | {compute_map3(act, pred):>8.5f}")
print("=" * 62)
```

```python
def expand_unitary_horizon_h24(
    vlm_probs: dict[str, float],
    bge_scores: dict[str, float],
    question_priors: dict[str, float],
    symbolic_bonuses: dict[str, float],
    horizon_k: int = 24,
) -> list[str]:
    """Construct the 24-candidate Unitary Horizon (H24) via weighted multi-stream fusion."""
    all_codes = set(vlm_probs) | set(bge_scores) | set(question_priors) | set(symbolic_bonuses) | {"NONE"}
    fused_priority: dict[str, float] = {}
    for code in all_codes:
        fused_priority[code] = (
            1.00 * vlm_probs.get(code, 0.0)
            + 0.45 * bge_scores.get(code, 0.0)
            + 0.35 * question_priors.get(code, 0.0)
            + 0.80 * symbolic_bonuses.get(code, 0.0)
        )
    ranked = sorted(fused_priority.items(), key=lambda kv: kv[1], reverse=True)
    shortlist = [code for code, _ in ranked[:horizon_k]]
    if "NONE" not in shortlist:
        shortlist[-1] = "NONE"
    return shortlist

# Demonstrate rescuing a true misconception (FRA-06) that was outside VLM Top-8
demo_vlm = {f"CAL-0{i}": 0.12 - i * 0.01 for i in range(1, 9)}
demo_bge = {"FRA-06": 0.78, "FRA-03": 0.54}
demo_priors = {"FRA-06": 0.62, "NONE": 0.35}
demo_sym = {"FRA-06": 1.15}

h24_pool = expand_unitary_horizon_h24(demo_vlm, demo_bge, demo_priors, demo_sym, horizon_k=10)
print("Top-5 Rescued Candidates in H24 Pool:", h24_pool[:5])
assert h24_pool[0] == "FRA-06", "FRA-06 must be rescued to Rank #1!"
```

  <p style="margin: 4px 0 0 0; font-size: 13px; color: #94a3b8;">How to re-rank 75 error codes freely while mathematically locking the 0.94246 Clean Gate boundary</p>

Let $s_{\text{anchor}}(i, c)$ be the anchor probability distribution for trace $i$. If we add auxiliary paradigm scores $\Delta(i, c) = \sum_{k=1}^{5} w_k z_k(i, c)$ directly to error candidates $c \neq \text{NONE}$, the maximum error score increases and overtakes $s_{\text{anchor}}(i, \text{NONE})$ on clean workings.

Our **Max-Preserving Orthogonal Projection ($\mathcal{P}^{\perp}$)** shifts all blended non-`NONE` logits by the exact scalar offset required to preserve the original error ceiling:

$$M_{\text{orig}}(i) = \max_{c \neq \text{NONE}} s_{\text{anchor}}(i, c), \qquad M_{\text{new}}(i) = \max_{c \neq \text{NONE}} \big(s_{\text{anchor}}(i, c) + \Delta(i, c)\big)$$

$$\hat{s}(i, c) = \begin{cases} s_{\text{anchor}}(i, \text{NONE}) & \text{if } c = \text{NONE} \\[4pt] s_{\text{anchor}}(i, c) + \Delta(i, c) + \big(M_{\text{orig}}(i) - M_{\text{new}}(i)\big) & \text{if } c \neq \text{NONE} \end{cases}$$

$$\implies \max_{c \neq \text{NONE}} \hat{s}(i, c) \equiv M_{\text{orig}}(i)$$

```python
def max_preserving_orthogonal_projection(
    anchor_scores: dict[str, float],
    paradigm_deltas: dict[str, float],
    weight: float = 0.35,
) -> dict[str, float]:
    """Re-rank non-NONE error codes while strictly preserving the Clean Gate margin."""
    projected = dict(anchor_scores)
    error_codes = [c for c in anchor_scores if c != "NONE"]
    if not error_codes:
        return projected

    orig_max = max(anchor_scores[c] for c in error_codes)
    blended = {c: anchor_scores[c] + weight * paradigm_deltas.get(c, 0.0) for c in error_codes}
    new_max = max(blended.values())

    # Exact orthogonal shift back to original maximum error envelope
    shift = orig_max - new_max
    for c in error_codes:
        projected[c] = blended[c] + shift
    return projected

# Proof on a borderline Clean trace where NONE=0.80 and KPF-02=0.74, KPF-01=0.69
anchor = {"NONE": 0.800, "KPF-02": 0.740, "KPF-01": 0.690, "CAL-01": 0.420}
aux_delta = {"KPF-01": 0.550, "KPF-02": 0.050, "CAL-01": -0.100}

naive_blend = {
    c: (anchor[c] if c == "NONE" else anchor[c] + 0.35 * aux_delta.get(c, 0.0))
    for c in anchor
}
orth_proj = max_preserving_orthogonal_projection(anchor, aux_delta, weight=0.35)

print("Naive Blend Top-3 (CORRUPTS NONE!):", sorted(naive_blend, key=naive_blend.get, reverse=True)[:3])
print("P^perp Proj Top-3 (PROTECTS NONE!):", sorted(orth_proj, key=orth_proj.get, reverse=True)[:3])
assert math.isclose(
    max(v for k, v in orth_proj.items() if k != "NONE"),
    max(v for k, v in anchor.items() if k != "NONE"),
    rel_tol=1e-9,
), "Max-Preserving Invariant Violated!"
print("✔ Mathematical Invariant Verified: max_{c!=NONE} s_hat(c) == 0.740000")
```
